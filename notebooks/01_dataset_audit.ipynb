```python
# Intent Discovery — AmericanAir

## Purpose

I manually inspected a random sample of 100 customer-to-AmericanAir interactions to understand the types of requests, complaints, follow-ups, and noise present in the dataset.

The goal of this stage was not to define the final taxonomy immediately, but to identify recurring customer goals and understand the ambiguity and noise in the real-world data.

The sample was drawn using a fixed random seed (`random_state=42`) from the AmericanAir customer-to-brand interactions.

---

## Initial observations

The sample contains a mixture of:

- Genuine customer support requests
- Complaints about service quality
- Follow-ups to existing conversations
- Requests involving multiple topics
- Simple acknowledgements or thank-you messages
- Social/reactional messages
- Ambiguous or low-information messages

This means that not every incoming tweet should necessarily be forced into a support intent.

For example:

- "What number can I call" represents a clear support request.
- "How can I fly standby if you cancelled my flight?" contains a clear actionable request.
- "Sent. Thanks." is an acknowledgement rather than a support request.
- "👍🏽" contains no actionable support request.
- "I'm still waiting..." may represent a follow-up to an existing case rather than a new support issue.

---

## Message-type observations

During manual inspection, I observed the following broad message types:

### 1. Support requests

Messages where the customer is asking AmericanAir to provide information, solve a problem, or take some action.

Examples observed:

- Asking how to fly standby after a cancelled flight.
- Asking for a contact number.
- Asking about a flight.
- Asking for a refund or compensation.
- Asking about changing or cancelling travel plans.

### 2. Complaints

Messages primarily expressing dissatisfaction with AmericanAir, its staff, policies, delays, or service.

Examples observed:

- Complaints about poor customer service.
- Complaints about gate agents.
- Complaints about delays.
- Complaints about baggage handling.
- Complaints about aircraft or onboard conditions.

A complaint can still contain an actionable support intent. Therefore, complaint should not automatically become the customer's primary intent.

### 3. Follow-ups / existing cases

Messages indicating that the customer is continuing an existing support interaction.

Examples observed:

- "I responded."
- "DM sent."
- "Still waiting."
- "Not what the attendant said."
- Asking what was sent in a previous DM.

These messages may require different handling from a new support request because the necessary context may exist in an earlier conversation.

### 4. Thanks / acknowledgement

Messages such as:

- "Thank you"
- "Just did. Thank you!"
- "Sent. Thanks."
- "#thankyou"

These do not represent a new customer support intent.

### 5. Unclear / social / low-information messages

Some tweets contain very little actionable information.

Examples include:

- "👍🏽"
- "Yay!"
- "Ahhhhh. Noted."
- General expressions of excitement about flying.
- Statements directed at other users rather than a direct support request.

These should not be artificially assigned to a domain intent.

---

# Candidate Support Topics

The following candidate topics repeatedly appeared during the manual inspection.

## 1. Flight disruption

Includes situations involving:

- Flight delays
- Flight cancellations
- Missed connections
- Reboarding
- Missing crew
- Gate-related disruption
- Operational delays
- Requests for alternative arrangements after disruption

Examples from the discovery sample include:

- Example 14 — customer wants to cancel without penalty because of expected flight disruption.
- Example 32 — customer asks about standby after a cancelled flight.
- Example 40 — customer has spent several hours at DFW and asks about compensation.
- Example 72 — customer is waiting for a first officer and experiencing a major delay.
- Example 95 — customer expects to miss a connection because of a delay.
- Example 100 — customer is concerned about making a flight because of delays.

---

## 2. Baggage

Includes:

- Missing baggage
- Delayed baggage
- Baggage tracking
- Baggage-related complaints
- Baggage fees
- Baggage refunds or compensation

Examples:

- Example 17 — customer has been waiting for a bag.
- Example 25 — customer says the bag is missing.
- Example 27 — customer compares a flight delay with a missing suitcase.
- Example 36 — customer is waiting for information regarding a baggage refund.
- Example 66 — customer has been waiting for luggage.

---

## 3. Booking / itinerary changes

Includes requests or problems involving:

- Changing a flight
- Cancelling a flight
- Rescheduling
- Same-day changes
- Changing travel plans
- Avoiding cancellation/change penalties

Examples:

- Example 3 — customer wants to reschedule a flight.
- Example 8 — customer questions a change fee after a natural disaster.
- Example 14 — customer wants to cancel without penalty.
- Example 90 — customer questions the same-day change policy.
- Example 94 — customer has an urgent trip the following day.

---

## 4. Seats / seat assignment

Includes:

- Seat assignment problems
- Paid seat issues
- Seats not being together
- Exit-row seats
- Seat availability
- Requests for seat changes

Examples:

- Example 6 — customer complains about a paid seat being reassigned.
- Example 62 — customer questions what they receive after paying for exit-row seats.
- Example 63 — customer complains about a first-class seat not working.
- Example 84 — customers had to request seats together.
- Example 99 — customer asks whether they have a guaranteed seat.

---

## 5. Upgrades / loyalty benefits

Includes:

- First-class upgrades
- Upgrade eligibility
- Priority boarding
- Loyalty status
- Status matching
- Loyalty benefits

Examples:

- Example 10 — customer complains about never receiving upgrades.
- Example 22 — customer complains about Priority boarding.
- Example 24 — customer questions priority boarding eligibility.
- Example 33 — customer requests a first-class upgrade.
- Example 37 — customer asks about Delta matching their status.
- Example 86 — customer says confirmed upgrades did not come through.

---

## 6. Fees / charges

Includes:

- Change fees
- Baggage fees
- Duplicate charges
- Unexpected charges
- Questions about penalties or fees

Examples:

- Example 8 — customer questions a $200 change fee.
- Example 52 — customer does not want to pay twice for the same ticket.
- Example 58 — customer complains about being charged for baggage.
- Example 65 — customer believes a fee was improperly charged.
- Example 85 — customer requests a refund.

---

## 7. Refunds / compensation

Includes requests for:

- Refunds
- Compensation after disruption
- Reimbursement
- Goodwill compensation

Examples:

- Example 36 — baggage refund.
- Example 40 — asks whether goodwill compensation is available.
- Example 63 — asks what AmericanAir will do to compensate them.
- Example 85 — explicitly requests a refund.

This category may overlap with fees, baggage, and flight disruption, so the boundary will need to be defined during taxonomy refinement.

---

## 8. Flight information / status

Includes requests for information about:

- Flight status
- Flight operation
- Flight schedules
- Whether a particular flight will operate
- Whether a customer has a confirmed seat

Examples:

- Example 39 — asks what is happening with flight 2289.
- Example 45 — asks about the return of a particular route.
- Example 72 — asks why the flight is delayed.
- Example 99 — asks about a guaranteed seat on flight 3059.

---

## 9. Airport / gate / staff issues

Includes problems involving:

- Gate agents
- Airport staff
- Boarding process
- Airport customer service
- Gate closures
- Airport-level assistance

Examples:

- Example 3 — customer complains about the airport team.
- Example 4 — gate was closed before the customer could board.
- Example 12 — complaint about flight staff.
- Example 41 — customer says the gate agent provided misleading information.
- Example 44 — complaint about the handling of the issue at Dallas headquarters.
- Example 56 — customer says airport/customer service did not help.
- Example 74 — customer says airport staff and supervisors did not help.

This topic may overlap heavily with flight disruption and boarding issues.

---

## 10. Onboard experience / aircraft issues

Includes:

- Power/electrical problems
- Cabin temperature
- Aircraft condition
- In-flight equipment
- Entertainment
- General onboard experience

Examples:

- Example 16 — no power available to use a computer or entertainment.
- Example 51 — complaint about aircraft type.
- Example 60 — complaint about aircraft/fleet condition.
- Example 79 — maintenance fixed an issue.
- Example 82 — complaint about cabin temperature and aisle conditions.

---

## 11. Contact / customer-service access

Includes requests or complaints involving:

- Contact information
- Phone support
- Difficulty reaching an agent
- Long telephone holds
- Requests to communicate privately through DM

Examples:

- Example 5 — asks what number to call.
- Example 46 — complains about lack of human contact information.
- Example 47 — asks what DM means.
- Example 54 — repeatedly placed on hold.
- Example 78 — compares Twitter response time with phone support.

---

# Important taxonomy observations

Several candidate topics overlap.

For example:

### Flight disruption vs. compensation

> "I've been delayed for five hours. Is there any good faith available?"

This contains both a disruption and a compensation request.

### Baggage vs. refund

> "Still waiting on the specific customer contact info for baggage refund."

This contains both baggage and refund.

### Airport issue vs. flight disruption

> "The gate agent lied and delayed us while waiting for another flight attendant."

This contains both an operational disruption and a complaint about airport staff.

Therefore, the final taxonomy should define a **primary-intent rule** rather than assigning multiple labels whenever possible.

---

# Preliminary Primary-Intent Rule

For the first taxonomy iteration, the primary intent should represent:

> **The customer's main actionable goal — what they want AmericanAir to help them accomplish.**

The emotional tone of the message should not determine the intent.

For example:

- "Your service is terrible. Where is my luggage?" → Baggage
- "I'm furious about the delay. Can I get a refund?" → Refund/compensation
- "Your gate agent was useless. Can you reschedule me?" → Booking/change
- "Thanks, I responded." → Follow-up/acknowledgement rather than a domain support intent

This rule will be tested and refined against additional examples before the final taxonomy is frozen.

---

# What I will not treat as a support intent

The following should not automatically become domain intents:

- Thanks / acknowledgement
- Generic praise
- Social reactions
- Messages with no actionable request
- Existing-case follow-ups when no new support goal is expressed
- Completely ambiguous messages

These cases may instead be handled by a separate message-type or routing layer.

---

# Next step

The candidate topics identified from the 100-example discovery sample are:

1. Flight disruption
2. Baggage
3. Booking / itinerary changes
4. Seats / seat assignment
5. Upgrades / loyalty benefits
6. Fees / charges
7. Refunds / compensation
8. Flight information / status
9. Airport / gate / staff issues
10. Onboard experience / aircraft issues
11. Contact / customer-service access

These are **candidate categories, not the final taxonomy**.

The next stage is to test these candidates against a larger sample and determine which categories should be:

- Kept as separate intents
- Merged
- Removed because they are too rare
- Treated as routing/handling conditions rather than intents
- Split because they contain meaningfully different customer goals
```

```python
# Taxonomy Validation — AmericanAir

## Purpose

I tested the initial candidate taxonomy against 25 previously unseen AmericanAir customer messages.

The purpose of this validation was to determine:

- Whether the candidate intents cover common customer requests.
- Whether different intents are clearly distinguishable.
- Where multiple intents appear in the same message.
- Which messages cannot be reliably classified using the current taxonomy.
- Whether additional intents are required before creating the final Golden Set.

The validation sample was drawn from AmericanAir customer-to-brand interactions that were not included in the original 100-message discovery sample.

---

## Candidate taxonomy

The initial candidate taxonomy consisted of:

- `flight_disruption`
- `baggage`
- `booking_change`
- `seat`
- `loyalty_upgrade`
- `fees_charges`
- `refund_compensation`
- `flight_information`
- `airport_gate_staff`
- `onboard_aircraft`
- `contact_support`
- `follow_up`
- `non_support`
- `other_unclear`

---

## Labeling rule

The primary intent represents the customer's main actionable goal.

When a message contains multiple issues, I select the intent corresponding to the customer's primary actionable request rather than assigning multiple domain intents.

The emotional tone of a message does not determine its intent.

For example:

- A customer can be angry but still have a `baggage` intent.
- A customer can complain about a delay while primarily requesting a `refund_compensation`.
- A customer can mention a fee while primarily requesting a `booking_change`.

Messages that do not contain a meaningful support request are classified as `non_support`, `follow_up`, or `other_unclear` depending on their role in the conversation.

---

# Validation Results — Examples 1–25

| # | Primary intent | Confidence | Notes |
|---:|---|---|---|
| 1 | `booking_change` | medium | Customer wants to cancel a flight; the $400 fee is an additional issue. |
| 2 | `other_unclear` | high | Trip insurance request is not represented by the current taxonomy. |
| 3 | `follow_up` | high | Customer indicates that they have sent a DM as part of an existing interaction. |
| 4 | `non_support` | high | Acknowledgement with no new support request. |
| 5 | `airport_gate_staff` | high | Customer is asking for help because no staff are available at the airport counters. |
| 6 | `airport_gate_staff` | high | Complaint and request related to the behavior of a gate agent. |
| 7 | `other_unclear` | low | Customer asks for an update but the subject of the update is not clear from the message. |
| 8 | `non_support` | high | Positive social statement with no support request. |
| 9 | `other_unclear` | high | Promotion/product question is not represented by the current taxonomy. |
| 10 | `other_unclear` | low | Context is insufficient to determine the underlying support intent. |
| 11 | `booking_change` | high | Customer wants to give up the original flight and obtain a new flight; the fee is part of the problem. |
| 12 | `onboard_aircraft` | medium | Primary complaint concerns aircraft facilities/condition, although a short delay is also mentioned. |
| 13 | `other_unclear` | medium | Customer only provides a record locator without stating the support goal. |
| 14 | `other_unclear` | medium | Transportation/Uber availability issue does not fit the current taxonomy clearly. |
| 15 | `loyalty_upgrade` | medium | Appears related to boarding-zone/priority treatment, but the exact customer request is unclear. |
| 16 | `fees_charges` | medium | Customer disputes baggage charges and also reports a paid exit-row seat not being provided. Multiple intents are present. |
| 17 | `refund_compensation` | high | Explicit recommendation/request for partial refund and food vouchers following a delay. |
| 18 | `non_support` | high | Positive acknowledgement with no actionable support request. |
| 19 | `other_unclear` | high | Customer is requesting a name-format/account-information change, which is not represented in the current taxonomy. |
| 20 | `non_support` | high | No actionable support request. |
| 21 | `baggage` | high | Customer is disputing baggage/check-in charges and baggage size requirements. |
| 22 | `follow_up` | medium | Appears to be a continuation of an existing support interaction, but context is limited. |
| 23 | `flight_disruption` | low | Customer appears to have missed/been unable to board a scheduled flight, but the exact cause and request are unclear. |
| 24 | `other_unclear` | high | Online check-in/technical glitch is the main issue, but a dedicated `check_in` intent does not currently exist. |
| 25 | `non_support` | high | Thank-you message with no new support request. |

---

# Taxonomy Gaps and Ambiguities

The validation sample revealed several limitations in the initial taxonomy.

## 1. Check-in is missing

Examples involving online check-in or check-in-related technical problems do not fit naturally into the existing taxonomy.

Example:

> Example 24 — customer reports that a technical glitch affected online check-in.

Potential new intent:

`check_in`

---

## 2. Insurance is missing

Example 2 asks whether trip insurance can be added after purchasing a ticket.

This represents a legitimate travel-support request but does not fit naturally into booking changes, fees, or flight information.

Potential new intent:

`insurance`

---

## 3. Promotions are missing

Example 9 concerns a Casper promotion and whether the promotion is branded correctly.

This suggests that promotional/offer-related questions may need their own category if they occur frequently enough in the broader dataset.

Potential new intent:

`promotions`

---

## 4. Account/customer-information requests are missing

Example 19 concerns the formatting of the customer's name.

This does not fit naturally into booking changes or flight information.

Potential new intent:

`account_information`

---

## 5. Transportation assistance may be outside the intended scope

Example 14 discusses whether an Uber option will be available.

This may be part of a broader travel-support issue but does not clearly belong to an AmericanAir-specific support intent.

This should be investigated on a larger sample before adding a dedicated category.

---

## 6. Multiple intents occur frequently

Some messages contain several legitimate issues.

For example, Example 16 contains:

- Baggage fee dispute
- TSA/precheck issue
- Paid exit-row seat problem

The taxonomy therefore requires a consistent primary-intent rule.

For this project, the selected rule is:

> Choose the customer's main actionable goal rather than assigning every topic mentioned in the message.

---

## 7. Context-dependent messages are common

Examples such as:

- "I DMd you"
- "They have tried repeatedly."
- "Have you heard of any updates?"
- "My record locator was BXMKZI"

cannot always be understood from the individual tweet alone.

This is important because the original dataset contains multi-turn conversations.

A production support agent should therefore consider available conversation history rather than relying only on a single tweet.

---

# Preliminary Taxonomy Decision

Based on this validation sample, the initial taxonomy should **not yet be frozen**.

The validation suggests that the following additional intents should be investigated:

- `check_in`
- `insurance`
- `promotions`
- `account_information`

At the same time, some existing categories may overlap and require further testing:

- `flight_disruption` vs. `airport_gate_staff`
- `fees_charges` vs. `refund_compensation`
- `flight_disruption` vs. `flight_information`
- `loyalty_upgrade` vs. boarding-related issues

The next step is to test the revised taxonomy on another unseen sample before freezing the final intent taxonomy.

---

# Next Step

The candidate taxonomy will be revised based on the observed gaps and then evaluated on another unseen batch of customer messages.

The final taxonomy will be frozen only after the revised categories provide reasonable coverage and the major sources of ambiguity have been documented.

The Golden Set will be created only after the taxonomy and labeling guidelines have been finalized.
```

```python
# Taxonomy Freeze

## Final Intent Taxonomy

The taxonomy was developed through an initial discovery sample followed by
unseen validation examples. The final taxonomy contains 15 categories.

| Intent | Definition |
|---|---|
| `flight_disruption` | Delays, cancellations, missed connections, deplaning, or other active flight disruptions. |
| `baggage` | Checked/carry-on baggage, lost bags, baggage fees, or baggage handling issues. |
| `booking_change` | Changing, cancelling, rebooking, or modifying an existing itinerary. |
| `seat` | Seat assignment, seating together, paid seats, or seat-related issues. |
| `loyalty_upgrade` | Loyalty status, upgrades, priority benefits, or upgrade-related issues. |
| `fees_charges` | Complaints or questions primarily about fees, charges, or additional costs. |
| `refund_compensation` | Refunds, compensation, vouchers, or reimbursement requests. |
| `flight_information` | Requests or comments primarily seeking flight, gate, schedule, or connection information. |
| `airport_gate_staff` | Issues involving airport/gate operations or employee/staff behavior. |
| `onboard_aircraft` | Issues concerning the aircraft or onboard experience. |
| `contact_support` | Attempts to contact customer support or questions about support channels. |
| `check_in` | Online/airport check-in problems or questions. |
| `follow_up` | A continuation of an existing support conversation without enough information to establish a new primary intent. |
| `non_support` | Thanks, reactions, casual comments, or other messages with no actionable support request. |
| `other_unclear` | A potentially support-related message that cannot be reliably assigned to another intent from the available context. |

## Labeling Rules

- Assign the **primary actionable intent**.
- Emotional tone does not determine the intent.
- For multiple issues, choose the customer's main actionable goal.
- Use `other_unclear` when the available message lacks enough context.
- Use `non_support` for messages with no actionable support request.
- Do not infer an intent that is not supported by the message.
```

```python
# Taxonomy v2 — AmericanAir

Based on the initial discovery sample and the first 25-example validation sample, I revised the candidate taxonomy.

The taxonomy is still provisional and will be tested on another unseen sample before being frozen.

## Proposed intents

### 1. flight_disruption
Issues involving delays, cancellations, missed connections, crew/operational disruptions, or other disruptions affecting the customer's flight.

### 2. baggage
Issues involving lost, delayed, damaged, checked, or incorrectly charged baggage.

### 3. booking_change
Requests to change, cancel, reschedule, or otherwise modify an existing flight booking.

### 4. seat
Issues involving seat assignment, paid seats, exit-row seats, seating arrangements, or seat availability.

### 5. loyalty_upgrade
Issues involving upgrades, loyalty status, priority boarding, or loyalty benefits.

### 6. fees_charges
Questions or disputes primarily concerning fees, charges, or penalties.

### 7. refund_compensation
Requests for refunds, reimbursement, compensation, vouchers, or goodwill compensation.

### 8. flight_information
Requests for flight information, schedules, routes, or status when the customer is not primarily reporting a disruption.

### 9. airport_gate_staff
Issues primarily involving airport staff, gate agents, boarding staff, or airport customer-service personnel.

### 10. onboard_aircraft
Issues involving the aircraft, cabin environment, onboard equipment, facilities, or onboard experience.

### 11. contact_support
Requests for phone numbers, contact information, or assistance reaching customer support.

### 12. check_in
Issues involving online check-in, check-in technical problems, or inability to complete the check-in process.

### 13. follow_up
A continuation of an existing support interaction without a clearly stated new support problem.

### 14. non_support
Messages that do not contain a new support request, such as thanks, praise, acknowledgements, or social reactions.

### 15. other_unclear
Messages where the customer's support goal cannot be reliably determined from the available context, or where the topic does not fit the current taxonomy.

---

## Categories not yet promoted to dedicated intents

The first validation sample contained isolated examples involving:

- Trip insurance
- Promotions
- Account/name information
- Transportation assistance

These are not yet promoted to dedicated intents because one or a few examples are insufficient evidence that they represent frequent enough support categories.

They will remain under `other_unclear` during the next validation round unless the broader sample shows that they occur frequently.

---

## Primary-intent rule

When multiple issues occur in one message, assign the intent corresponding to the customer's primary actionable goal.

Do not assign multiple domain intents.

If no meaningful support goal is present, use `non_support`, `follow_up`, or `other_unclear` as appropriate.

---

## Important ambiguity boundaries

### flight_disruption vs. flight_information

Use `flight_disruption` when something has gone wrong or is actively disrupting the customer's travel.

Use `flight_information` when the customer is primarily requesting information without reporting a disruption.

### fees_charges vs. refund_compensation

Use `fees_charges` when the main issue is a fee or charge.

Use `refund_compensation` when the customer is primarily asking to receive money, reimbursement, vouchers, or compensation.

### flight_disruption vs. airport_gate_staff

Use `airport_gate_staff` when the main problem is the behavior or action of airport/gate personnel.

Use `flight_disruption` when the main problem is the operational disruption itself.

### seat vs. loyalty_upgrade

Use `seat` for a specific seat assignment or paid-seat problem.

Use `loyalty_upgrade` for upgrades, status, priority boarding, or loyalty benefits.

---

## Taxonomy status

This taxonomy is provisional.

It will be tested against another unseen sample before being frozen for Golden Set construction.
```

```python
golden_pool = golden_pool.copy()

# Create message-length groups
golden_pool["length_group"] = pd.qcut(
    golden_pool["customer_message"].str.len(),
    q=4,
    labels=["short", "medium", "long", "very_long"],
    duplicates="drop"
)

# Sample proportionally across message-length groups
golden_set = (
    golden_pool
    .groupby("length_group", observed=True, group_keys=False)
    .apply(
        lambda x: x.sample(
            n=round(200 * len(x) / len(golden_pool)),
            random_state=42
        )
    )
)

# Ensure exactly 200 examples
if len(golden_set) > 200:
    golden_set = golden_set.sample(n=200, random_state=42)

elif len(golden_set) < 200:
    remaining = golden_pool.loc[
        ~golden_pool.index.isin(golden_set.index)
    ]

    golden_set = pd.concat([
        golden_set,
        remaining.sample(
            n=200 - len(golden_set),
            random_state=42
        )
    ])

# Recreate length_group because groupby/apply may remove it
golden_set["length_group"] = pd.qcut(
    golden_set["customer_message"].str.len(),
    q=4,
    labels=["short", "medium", "long", "very_long"],
    duplicates="drop"
)

print("Golden Set size:", len(golden_set))
print("\nLength distribution:")
print(golden_set["length_group"].value_counts())print("Discovery overlap:",
      len(set(golden_set["customer_tweet_id"]) &
          set(discovery_sample["customer_tweet_id"])))

print("V1 validation overlap:",
      len(set(golden_set["customer_tweet_id"]) &
          set(taxonomy_validation["customer_tweet_id"])))

print("V2 validation overlap:",
      len(set(golden_set["customer_tweet_id"]) &
          set(taxonomy_validation_2["customer_tweet_id"])))
```
